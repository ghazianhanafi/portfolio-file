```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML

# =====================================================================
# 1. MATHEMATICAL SETUP
# =====================================================================
# We evaluate x from 0 to 115 because the crossover happens around x=95.4
x = np.linspace(0, 115, 250)
y1 = x**2
y2 = 1.1**x

# Find the exact index where 1.1^x overtakes x^2 (ignoring the tiny blip at x=1)
crossover_idx = np.where((y2 > y1) & (x > 10))[0][0]
crossover_x = x[crossover_idx]
crossover_y = y2[crossover_idx]

# =====================================================================
# 2. FIGURE & AESTHETICS SETUP
# =====================================================================
# Use a dark background for dramatic effect
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(10, 6))

ax.set_xlim(0, 115)
ax.set_ylim(0, 500) # This will dynamically scale up

# Create the line objects
line1, = ax.plot([], [], lw=3, color='#00d2ff', label='x² (Polynomial)')
line2, = ax.plot([], [], lw=3, color='#a0a0a0', label='1.1ˣ (Exponential)')

# Dramatic mark elements
explosion, = ax.plot([], [], marker='*', color='yellow', markersize=20, alpha=0)
annotation = ax.text(0.5, 0.7, '', transform=ax.transAxes,
                     fontsize=24, color='#ff3333', fontweight='bold',
                     ha='center', va='center')

ax.legend(loc='upper left', fontsize=12)
ax.set_title("The Illusion of Linear Thinking: x² vs 1.1ˣ", fontsize=16, pad=15)
ax.set_xlabel("x", fontsize=14)
ax.set_ylabel("y", fontsize=14)
ax.grid(True, linestyle=':', alpha=0.3)

# =====================================================================
# 3. ANIMATION LOGIC
# =====================================================================
def init():
    line1.set_data([], [])
    line2.set_data([], [])
    explosion.set_data([], [])
    annotation.set_text('')
    return line1, line2, explosion, annotation

def update(frame):
    # Get data up to the current frame
    curr_x = x[:frame]
    curr_y1 = y1[:frame]
    curr_y2 = y2[:frame]

    line1.set_data(curr_x, curr_y1)
    line2.set_data(curr_x, curr_y2)

    # Dynamically scale the Y-axis to give a sense of extreme acceleration
    if frame > 0:
        current_max_y = max(max(curr_y1), max(curr_y2))
        ax.set_ylim(0, max(500, current_max_y * 1.05))

    # 💥 THE DRAMATIC TRIGGER 💥
    if frame >= crossover_idx:
        # Exponential line turns bright fiery red and gets thicker
        line2.set_color('#ff3333')
        line2.set_linewidth(4)

        # Show the explosion star at the crossover point
        explosion.set_data([crossover_x], [crossover_y])
        explosion.set_alpha(1.0)

        # Flashing text effect (blinks every 8 frames)
        if (frame // 8) % 2 == 0:
            annotation.set_text("EXPONENTIAL TAKEOVER!\n1.1ˣ > x²")
        else:
            annotation.set_text("")
    else:
        # Reset elements if animation loops
        line2.set_color('#a0a0a0')
        line2.set_linewidth(3)
        explosion.set_alpha(0)
        annotation.set_text("")

    return line1, line2, explosion, annotation

# Create the animation
anim = FuncAnimation(fig, update, frames=len(x), init_func=init,
                     blit=False, interval=60)

# =====================================================================
# 4. EXPORT & RENDER
# =====================================================================
# To save as a GIF to your local machine (uncomment to run):
# print("Saving GIF...")
# anim.save('exponential_takeover.gif', writer=PillowWriter(fps=25))
# print("GIF Saved!")

# Render interactively inside the Jupyter Notebook
plt.close() # Prevents duplicate static plot from showing
HTML(anim.to_jshtml())
```
